```python
import kagglehub
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, precision_score, recall_score, f1_score
from imblearn.over_sampling import SMOTE
from imblearn.under_sampling import TomekLinks
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
```

```python
path = kagglehub.dataset_download("mlg-ulb/creditcardfraud")
```

```python
os.listdir(path)
```

## Assessing the Data

```python
df = pd.read_csv(os.path.join(path, 'creditcard.csv'))
```

```python
#Transform the amount variable to log amount to prevent large change of scale
# normalizing the class label counts to set the heights of the bars and the displays
counts = df['Class'].value_counts(normalize=True) * 100
counts[1] = counts[1] + 0.6 #esnuring that the bar for the minority class is visible
```

```python
#Creating a bar plot
plt.bar(x = df['Class'].unique(), height = counts, color = (0.8,0.5,0), tick_label = ['Legitimate', 'Fraudulent'])
plt.title('Percentage of Transactions in each Class')
plt.ylim(0,105)
plt.text(x = -0.05, y = 60, s = "{0:5.3f}%".format(counts[0]))
plt.text(x = 0.95, y = 10, s = "{0:4.3f}%".format(counts[1] - 0.6))
plt.show()
```

```python
y = df['Class']
X = df.drop(columns = ['Class', 'Time'])
```

## Train Test Split, SMOTE and TomekLinks

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
#SMOTE --> Synthetic Minority Over-sampling Technique
smote = SMOTE(random_state=42)
X_trained_resampled, y_trained_resampled = smote.fit_resample(X_train, y_train)
```

```python
#Tomeklinks --> Cleans up borderline cases introduced by SMOTE
tl = TomekLinks()
X_trained_resampled, y_trained_resampled = tl.fit_resample(X_trained_resampled, y_trained_resampled)
```

### Scaling

```python
#Scaling
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_trained_resampled)
X_test_scaled = scaler.transform(X_test)
```

## Logistics Regression

```python
#Applying logistic regression
lgr = LogisticRegression(penalty='l2', C=1.0, max_iter=1000) #Ridged regularisation
lgr.fit(X_train_scaled, y_trained_resampled)
y_pred_lgr = lgr.predict(X_test_scaled)
```

```python
classification_lgr = classification_report(y_test, y_pred_lgr)
confusion_lgr = confusion_matrix(y_test, y_pred_lgr)
f1_lgr = f1_score(y_test, y_pred_lgr)
precision_lgr = precision_score(y_test, y_pred_lgr)
recall_lgr = recall_score(y_test, y_pred_lgr)

print(f"Classification report for Logistics Regression: {classification_lgr}")
print(f"Confusion matrix for Logistics Regression: {confusion_lgr}")
print(f"F1 score for Logistics Regression: {f1_lgr:.4f}")
print(f"Precision score for Logistics Regression: {precision_lgr:.4f}")
print(f"Recall score for Logistics Regression: {recall_lgr:.4f}")
```

```python
crf = RandomForestClassifier(n_estimators=100, min_samples_split=2, max_depth=15)
crf.fit(X_train_scaled, y_trained_resampled)
y_pred_crf = crf.predict(X_test_scaled)
```

```python
classification_crf = classification_report(y_test, y_pred_crf)
confusion_crf = confusion_matrix(y_test, y_pred_crf)
f1_crf = f1_score(y_test, y_pred_crf)
precision_crf = precision_score(y_test, y_pred_crf)
recall_crf = recall_score(y_test, y_pred_crf)

print(f"Classification report for Logistics Regression: {classification_crf}")
print(f"Confusion matrix for Logistics Regression: {confusion_crf}")
print(f"F1 score for Logistics Regression: {f1_crf:.4f}")
print(f"Precision score for Logistics Regression: {precision_crf:.4f}")
print(f"Recall score for Logistics Regression: {recall_crf:.4f}")
```
